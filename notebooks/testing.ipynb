```python
import pyspiel
from transformers import pipeline
```

```python
print(pyspiel.registered_names())
```

```python
game = pyspiel.load_game("tic_tac_toe")
```

```python
print(game)
```

```python
print(game.num_players())
```

```python
state = game.new_initial_state()
```

```python
print(state.current_player())
print(state.legal_actions())

for action in state.legal_actions():
    print(f"{action} {state.action_to_string(action)}")
```

```python
print(f"You are the first player in a 3x3 game of tic tac toe. Your available actions are: {[state.action_to_string(action) for action in state.legal_actions()]}")
```

```python
game_player = pipeline(task="question-answering", model="deepset/roberta-base-squad2")

move = game_player(
    question="of the available actions in a 3x3 game of tic-tac-toe, to choose a location with the format (x,y). you are the computer player, playing against a human.",
    context=f"available actions:{[state.action_to_string(action) for action in state.legal_actions()]}\n\ncomputer locations"
)

print(
    f"score: {round(move['score'], 4)}, start: {move['start']}, end: {move['end']}, answer: {move['answer']}"
)
```
